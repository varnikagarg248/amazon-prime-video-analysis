# **Project Summary -**

The Amazon Prime TV Shows and Movies project is an Exploratory Data Analysis (EDA) project designed to analyze the content library of Amazon Prime Video and identify meaningful patterns that can support data-driven content and business decisions. The analysis uses two datasets, titles.csv and credits.csv, containing information about titles, content types, genres, release years, production countries, ratings, popularity metrics, and the actors and directors associated with each title. The project follows a structured Univariate, Bivariate, and Multivariate (UBM) approach to understand individual variables as well as relationships between multiple variables.

The first stage of the project focuses on understanding and preparing the data for analysis. Data wrangling was performed to improve the quality, consistency, and reliability of the datasets. Duplicate records were identified and removed, while missing values were systematically analyzed across descriptions, age certifications, seasons, IMDb information, and TMDB metrics. Text-based missing values were handled using meaningful placeholders, while categorical fields such as age certification were assigned an "Unknown" category. Season information was treated according to content type, with values that were not applicable to Movies distinguished from genuinely missing information. An additional feature, imdb_missing_info, was created to record missing IMDb-related attributes and provide greater visibility into data completeness. Numerical missing values such as IMDb score, IMDb votes, TMDB popularity, and TMDB score were handled using group-based median imputation based on content type and age certification, with content-type medians used as a fallback.

The exploratory analysis examines Amazon Prime's content catalogue from multiple perspectives. Genre analysis identifies the most represented and underrepresented genres, while the Movie-versus-TV-Show analysis examines the overall composition of the catalogue and differences in content-type representation across genres. The project also analyzes cast size and the genre associations of leading actors and directors, providing insights into talent requirements and specialization. Production geography is explored through production-country contributions, international production diversity, IMDb rating ranges by country, and talent density across different production volumes.

The project further investigates content production trends over time, comparing the growth of Movies and TV Shows across individual years and broader time periods. IMDb ratings are analyzed through distributions, box plots, highly rated content trends, and rankings of top-rated titles with sufficient IMDb votes. Multivariate analysis uses correlation heatmaps and pair plots to examine relationships among runtime, seasons, IMDb scores, IMDb votes, TMDB popularity, and TMDB scores.

Overall, the analysis provides Amazon Prime with insights into content diversity, genre trends, production patterns, content quality, geographical contribution, and talent involvement. These insights can support better content acquisition and production decisions by identifying successful content areas, underrepresented genres and regions, and experienced actors and directors. A data-driven approach can help Amazon Prime strengthen content quality, improve audience engagement, diversify its catalogue, and support sustainable long-term growth.

## 3. ***Data Wrangling***



### Data Wrangling Code

````python
titles_df.isna().sum()
````

````python
# ==========================================================
# STEP 1 : Remove Duplicate Records
# ==========================================================

print("\n========== DUPLICATE RECORD REMOVAL ==========\n")

duplicate_count = titles_df.duplicated().sum()

print(f"Duplicate Records Before Cleaning : {duplicate_count}")

titles_df.drop_duplicates(inplace=True)

print(f"Duplicate Records After Cleaning  : {titles_df.duplicated().sum()}")

print("\nDuplicate records removed successfully.")
titles_df.isna().sum()
````

````python
# ==========================================================
# STEP 2 : Missing Values Before Data Cleaning
# ==========================================================

print("\n========== MISSING VALUES BEFORE CLEANING ==========\n")

print(titles_df.isnull().sum())
````

````python
# ==========================================================
# STEP 3 : Handle Missing Description Values
# ==========================================================

print("\n========== DESCRIPTION VALUE CLEANING ==========\n")

missing_description = titles_df["description"].isnull().sum()

print(f"Missing Descriptions Before Cleaning : {missing_description}")


titles_df["description"] = titles_df["description"].fillna("No Description Available")


print(
    f"Missing Descriptions After Cleaning  : "
    f"{titles_df['description'].isnull().sum()}"
)

print("\nDescription values handled successfully.")
print(titles_df.isnull().sum())
````

````python
# ==========================================================
# STEP 4 : Handle Missing Age Certification Values
# ==========================================================

print("\n========== AGE CERTIFICATION CLEANING ==========\n")

missing_age = titles_df["age_certification"].isnull().sum()

print(f"Missing Age Certifications Before Cleaning : {missing_age}")


titles_df["age_certification"] = titles_df["age_certification"].fillna("Unknown")


print(
    f"Missing Age Certifications After Cleaning  : "
    f"{titles_df['age_certification'].isnull().sum()}"
)

print("\nAge certification values handled successfully.")
print(titles_df.isnull().sum())
````

````python
# ==========================================================
# STEP 5 : Handle Missing Season Values
# ==========================================================

print("\n========== SEASON VALUE CLEANING ==========\n")

missing_seasons_before = titles_df["seasons"].isnull().sum()

print(f"Missing Seasons Before Cleaning : {missing_seasons_before}")


titles_df.loc[
    (titles_df["type"] == "MOVIE") &
    (titles_df["seasons"].isna()),
    "seasons"
] = "Not Applicable"


missing_seasons_after = titles_df["seasons"].isnull().sum()

print(f"Missing Seasons After Cleaning  : {missing_seasons_after}")

print("\nMovie season values updated successfully.")
print(titles_df.isnull().sum())
````

````python
# ==========================================================
# STEP 6 : Create IMDb Missing Information Feature
# ==========================================================

print("\n========== IMDb MISSING INFORMATION ANALYSIS ==========\n")


def missing_imdb_columns(row):
    missing = []

    if pd.isna(row["imdb_id"]):
        missing.append("imdb_id")

    if pd.isna(row["imdb_score"]):
        missing.append("imdb_score")

    if pd.isna(row["imdb_votes"]):
        missing.append("imdb_votes")

    return ", ".join(missing) if missing else "None"


titles_df["imdb_missing_info"] = titles_df.apply(
    missing_imdb_columns,
    axis=1
)


print("IMDb missing information column created successfully.")

print(
    titles_df["imdb_missing_info"].value_counts()
)
print(titles_df.isnull().sum())
````

````python
# ==========================================================
# STEP 7 : Handle Missing IMDb Score Values
# ==========================================================

print("\n========== IMDb SCORE VALUE IMPUTATION ==========\n")


# Check missing values before cleaning
missing_before = titles_df["imdb_score"].isnull().sum()

print(f"Missing IMDb Scores Before Cleaning : {missing_before}")


# Replace missing IMDb scores using the median IMDb score
# of the same content type and age certification

titles_df["imdb_score"] = (
    titles_df["imdb_score"]
    .fillna(
        titles_df.groupby(
            ["type", "age_certification"]
        )["imdb_score"]
        .transform("median")
    )
)


titles_df["imdb_score"] = (
    titles_df["imdb_score"]
    .fillna(
        titles_df.groupby("type")["imdb_score"]
        .transform("median")
    )
)


# Check missing values after cleaning
missing_after = titles_df["imdb_score"].isnull().sum()

print(f"Missing IMDb Scores After Cleaning  : {missing_after}")


print("\nIMDb score values handled successfully.")
print(titles_df.isnull().sum())
````

````python
# ==========================================================
# STEP 8 : Handle Missing IMDb Vote Values
# ==========================================================

print("\n========== IMDb VOTES VALUE IMPUTATION ==========\n")


# Check missing values before cleaning
missing_before = titles_df["imdb_votes"].isnull().sum()

print(f"Missing IMDb Votes Before Cleaning : {missing_before}")


# Replace missing IMDb votes using the median vote count
# of the same content type and age certification

titles_df["imdb_votes"] = (
    titles_df["imdb_votes"]
    .fillna(
        titles_df.groupby(
            ["type", "age_certification"]
        )["imdb_votes"]
        .transform("median")
    )
)


# Replace remaining missing values using
# the median vote count of the same content type

titles_df["imdb_votes"] = (
    titles_df["imdb_votes"]
    .fillna(
        titles_df.groupby("type")["imdb_votes"]
        .transform("median")
    )
)


# Check missing values after cleaning
missing_after = titles_df["imdb_votes"].isnull().sum()

print(f"Missing IMDb Votes After Cleaning  : {missing_after}")


print("\nIMDb vote values handled successfully.")
print(titles_df.isnull().sum())
````

````python
# ==========================================================
# STEP 9 : Handle Missing TMDB Popularity Values
# ==========================================================

print("\n========== TMDB POPULARITY VALUE IMPUTATION ==========\n")


# Check missing values before cleaning
missing_before = titles_df["tmdb_popularity"].isnull().sum()

print(f"Missing TMDB Popularity Before Cleaning : {missing_before}")


# Replace missing TMDB popularity using the median popularity
# of the same content type and age certification

titles_df["tmdb_popularity"] = (
    titles_df["tmdb_popularity"]
    .fillna(
        titles_df.groupby(
            ["type", "age_certification"]
        )["tmdb_popularity"]
        .transform("median")
    )
)


# Replace remaining missing values using
# the median popularity of the same content type

titles_df["tmdb_popularity"] = (
    titles_df["tmdb_popularity"]
    .fillna(
        titles_df.groupby("type")["tmdb_popularity"]
        .transform("median")
    )
)


# Check missing values after cleaning
missing_after = titles_df["tmdb_popularity"].isnull().sum()

print(f"Missing TMDB Popularity After Cleaning  : {missing_after}")


print("\nTMDB popularity values handled successfully.")
print(titles_df.isnull().sum())
````

````python
# ==========================================================
# STEP 10 : Handle Missing TMDB Score Values
# ==========================================================

print("\n========== TMDB SCORE VALUE IMPUTATION ==========\n")


# Check missing values before cleaning
missing_before = titles_df["tmdb_score"].isnull().sum()

print(f"Missing TMDB Scores Before Cleaning : {missing_before}")


# Replace missing TMDB scores using the median score
# of the same content type and age certification

titles_df["tmdb_score"] = (
    titles_df["tmdb_score"]
    .fillna(
        titles_df.groupby(
            ["type", "age_certification"]
        )["tmdb_score"]
        .transform("median")
    )
)


# Replace remaining missing values using
# the median score of the same content type

titles_df["tmdb_score"] = (
    titles_df["tmdb_score"]
    .fillna(
        titles_df.groupby("type")["tmdb_score"]
        .transform("median")
    )
)


# Check missing values after cleaning
missing_after = titles_df["tmdb_score"].isnull().sum()

print(f"Missing TMDB Scores After Cleaning  : {missing_after}")


print("\nTMDB score values handled successfully.")
print(titles_df.isnull().sum())
````

````python
# ==========================================================
# STEP 11 : Handle Missing IMDb ID Values
# ==========================================================

print("\n========== IMDb ID VALUE CLEANING ==========\n")


# Check missing values before cleaning
missing_before = titles_df["imdb_id"].isnull().sum()

print(f"Missing IMDb IDs Before Cleaning : {missing_before}")


# Replace missing IMDb IDs with a placeholder value

titles_df["imdb_id"] = (
    titles_df["imdb_id"]
    .fillna("ID Not Available")
)


# Check missing values after cleaning
missing_after = titles_df["imdb_id"].isnull().sum()

print(f"Missing IMDb IDs After Cleaning  : {missing_after}")


print("\nIMDb ID values handled successfully.")
print(titles_df.isnull().sum())
````

### What all manipulations have you done and insights you found?

The dataset underwent data cleaning and preprocessing process to improve data quality, consistency, and reliability before performing further analysis.

Initially, duplicate records were identified and removed to ensure that each title appeared only once in the dataset. This helped eliminate redundant information and prevented duplicate entries from influencing analytical results.

A missing value analysis was performed to understand the completeness of the dataset. Missing values were identified across different attributes, including descriptions, age certifications, season information, IMDb details, and TMDB ratings.

For text-based attributes such as descriptions, missing values were replaced with meaningful placeholders instead of removing records. This approach preserved the available data while clearly indicating unavailable information. Similarly, missing age certification values were replaced with an "Unknown" category to maintain consistency in categorical analysis.

The season information column was further cleaned by identifying movie records where season details were not applicable. These missing values were replaced with "Not Applicable" to differentiate between unavailable information and attributes that do not apply to a particular content type.

To improve data quality analysis, a new feature named imdb_missing_info was created. This feature captures which IMDb-related attributes were missing for each title, including IMDb ID, IMDb score, and IMDb vote count. This provided better visibility into missing data patterns and helped evaluate the reliability of IMDb information.

For numerical attributes such as IMDb score, IMDb votes, TMDB popularity, and TMDB score, missing values were handled using group-based median imputation. Missing values were first replaced using the median value calculated from similar content based on type and age certification. Remaining missing values were filled using the median value of the respective content type. This method ensured that replaced values maintained the characteristics of similar titles rather than using a generalized dataset average.

Finally, missing IMDb IDs were replaced with a placeholder value to maintain dataset completeness and avoid losing records during analysis.

Through these preprocessing steps, the dataset was transformed into a cleaner and more consistent format, making it suitable for exploratory data analysis, visualization, and further machine learning applications.

#### Chart - 1 - Horizontal Bar Chart

````python
genre_df = titles_df.copy()

genre_df['genres'] = genre_df['genres'].apply(
    lambda x: ast.literal_eval(x) if isinstance(x, str) else x
)

genre_df = genre_df.explode('genres')

genre_df['genres'] = genre_df['genres'].str.strip().str.capitalize()

genre_counts = genre_df['genres'].value_counts()

plt.figure(figsize=(12, 7))

ax = sns.barplot(
    x=genre_counts.values,
    y=genre_counts.index
)

# Title and labels
plt.title(
    'Amazon Prime Video Genre Distribution',
    fontsize=16,
    fontweight='bold',
    pad=15
)

plt.xlabel('Number of Titles', fontsize=12)
plt.ylabel('Genre', fontsize=12)

# Add values at the end of each bar
for i, value in enumerate(genre_counts.values):
    ax.text(
        value + 20,
        i,
        f'{value:,}',
        va='center',
        fontsize=10
    )

# Styling
ax.grid(axis='x', linestyle='--', alpha=0.3)
ax.grid(axis='y', visible=False)

plt.tight_layout()
plt.show()
````

#### Chart - 5 - Stacked Bar Chart

````python
credits_chart = credits_df.copy()
titles_chart = titles_df[['id', 'genres']].copy()

credits_chart['role'] = (
    credits_chart['role']
    .astype(str)
    .str.strip()
    .str.upper()
)

titles_chart['genres'] = titles_chart['genres'].apply(
    lambda x: ast.literal_eval(x)
    if isinstance(x, str)
    else x
)

titles_chart = titles_chart.dropna(
    subset=['genres']
)

titles_chart = titles_chart.explode('genres')

titles_chart['genres'] = (
    titles_chart['genres']
    .astype(str)
    .str.strip()
    .str.lower()
)

top_genres = (
    titles_chart['genres']
    .value_counts()
    .head(6)
    .index
    .tolist()
)

credits_genre = credits_chart.merge(
    titles_chart[['id', 'genres']],
    on='id',
    how='inner'
)

def prepare_role_data(role):

    role_data = credits_genre[
        credits_genre['role'] == role
    ].copy()

    top_people = (
        role_data
        .groupby(['person_id', 'name'])['id']
        .nunique()
        .sort_values(ascending=False)
        .head(5)
        .reset_index()
    )

    person_ids = top_people['person_id'].tolist()

    role_data = role_data[
        role_data['person_id'].isin(person_ids)
    ].copy()

    role_data['genre_group'] = role_data['genres'].apply(
        lambda x: x if x in top_genres else 'other'
    )

    counts = (
        role_data
        .groupby(
            ['person_id', 'name', 'genre_group']
        )['id']
        .nunique()
        .reset_index(name='title_count')
    )

    pivot = (
        counts
        .pivot_table(
            index=['person_id', 'name'],
            columns='genre_group',
            values='title_count',
            aggfunc='sum',
            fill_value=0
        )
        .reset_index()
    )

    for genre in top_genres + ['other']:
        if genre not in pivot.columns:
            pivot[genre] = 0

    pivot['total'] = pivot[
        top_genres + ['other']
    ].sum(axis=1)

    pivot = pivot.sort_values(
        'total',
        ascending=True
    )

    return pivot

actor_pivot = prepare_role_data('ACTOR')
director_pivot = prepare_role_data('DIRECTOR')


fig, axes = plt.subplots(
    1,
    2,
    figsize=(18, 8)
)

ax = axes[0]

left = pd.Series(
    0,
    index=actor_pivot.index
)

for genre in top_genres + ['other']:

    values = actor_pivot[genre]

    ax.barh(
        actor_pivot['name'],
        values,
        left=left,
        height=0.62,
        label=genre.title()
    )

    for i, (start, value) in enumerate(
        zip(left, values)
    ):

        if value >= 3:

            ax.text(
                start + value / 2,
                i,
                f'{int(value)}',
                ha='center',
                va='center',
                fontsize=8.5,
                fontweight='bold'
            )

    left += values


for i, (_, row) in enumerate(
    actor_pivot.iterrows()
):

    ax.text(
        row['total'] + 1,
        i,
        f"{int(row['total'])}",
        va='center',
        fontsize=10,
        fontweight='bold'
    )


ax.set_title(
    'Top 5 Actors',
    fontsize=16,
    fontweight='bold',
    pad=15
)

ax.set_xlabel(
    'Unique Title-Genre Associations',
    fontsize=11
)

ax.set_ylabel('')

ax.grid(
    axis='x',
    linestyle='--',
    linewidth=0.8,
    alpha=0.25
)

ax.grid(
    axis='y',
    visible=False
)

ax.set_axisbelow(True)

sns.despine(
    ax=ax,
    left=True,
    bottom=True
)

ax.tick_params(
    axis='y',
    labelsize=10
)

ax.tick_params(
    axis='x',
    labelsize=9
)

ax = axes[1]

left = pd.Series(
    0,
    index=director_pivot.index
)

for genre in top_genres + ['other']:

    values = director_pivot[genre]

    ax.barh(
        director_pivot['name'],
        values,
        left=left,
        height=0.62,
        label=genre.title()
    )

    for i, (start, value) in enumerate(
        zip(left, values)
    ):

        if value >= 3:

            ax.text(
                start + value / 2,
                i,
                f'{int(value)}',
                ha='center',
                va='center',
                fontsize=8.5,
                fontweight='bold'
            )

    left += values


for i, (_, row) in enumerate(
    director_pivot.iterrows()
):

    ax.text(
        row['total'] + 1,
        i,
        f"{int(row['total'])}",
        va='center',
        fontsize=10,
        fontweight='bold'
    )


ax.set_title(
    'Top 5 Directors',
    fontsize=16,
    fontweight='bold',
    pad=15
)

ax.set_xlabel(
    'Unique Title-Genre Associations',
    fontsize=11
)

ax.set_ylabel('')

ax.grid(
    axis='x',
    linestyle='--',
    linewidth=0.8,
    alpha=0.25
)

ax.grid(
    axis='y',
    visible=False
)

ax.set_axisbelow(True)

sns.despine(
    ax=ax,
    left=True,
    bottom=True
)

ax.tick_params(
    axis='y',
    labelsize=10
)

ax.tick_params(
    axis='x',
    labelsize=9
)

fig.suptitle(
    'Genre Associations of Top Actors and Directors',
    fontsize=20,
    fontweight='bold',
    y=1.02
)

fig.text(
    0.5,
    0.96,
    'Unique title-genre associations; individual titles may belong to multiple genres',
    ha='center',
    fontsize=11
)

handles, labels = axes[0].get_legend_handles_labels()

fig.legend(
    handles,
    labels,
    title='Genre',
    loc='lower center',
    bbox_to_anchor=(0.5, -0.02),
    ncol=4,
    frameon=False
)

plt.tight_layout(
    rect=[0, 0.07, 1, 0.94]
)

plt.show()
````

A stacked bar chart was selected to compare the genre associations of the top actors and directors. It shows both their total title associations and how these are distributed across different genres.

##### 2. What is/are the insight(s) found from the chart?

#### Chart - 15 - Horizontal Bar Chart

````python
#### Chart - 15 - Horizontal Bar Chart

top_rated = titles_df[
    ['id', 'title', 'type', 'imdb_score', 'imdb_votes']
].dropna(
    subset=['imdb_score', 'imdb_votes']
).copy()

top_rated['imdb_score'] = pd.to_numeric(
    top_rated['imdb_score'],
    errors='coerce'
)

top_rated['imdb_votes'] = pd.to_numeric(
    top_rated['imdb_votes'],
    errors='coerce'
)

top_rated = top_rated.dropna(
    subset=['imdb_score', 'imdb_votes']
)

MIN_VOTES = 1000

top_rated = top_rated[
    top_rated['imdb_votes'] >= MIN_VOTES
]

top_rated = top_rated.sort_values(
    ['imdb_score', 'imdb_votes'],
    ascending=[False, False]
).head(10)


# --------------------------------------------------
# Prepare credits
# --------------------------------------------------

credits_clean = credits_df[
    ['id', 'name', 'role']
].copy()

credits_clean['role'] = (
    credits_clean['role']
    .astype(str)
    .str.strip()
    .str.upper()
)

credits_clean['name'] = (
    credits_clean['name']
    .astype(str)
    .str.strip()
)


# --------------------------------------------------
# Collect all directors for each title
# --------------------------------------------------

directors = (
    credits_clean[
        credits_clean['role'] == 'DIRECTOR'
    ]
    .groupby('id')['name']
    .apply(
        lambda x: ', '.join(
            pd.unique(x)
        )
    )
    .reset_index()
    .rename(
        columns={'name': 'director'}
    )
)


# --------------------------------------------------
# Collect actors
# Show first 4 actors + remaining count
# --------------------------------------------------

def format_actors(names):

    names = list(
        pd.unique(names)
    )

    first_four = names[:4]
    remaining = len(names) - 4

    actor_text = ', '.join(
        first_four
    )

    if remaining > 0:
        actor_text += (
            f' + {remaining} '
            f'{"actor" if remaining == 1 else "actors"}'
        )

    return actor_text


actors = (
    credits_clean[
        credits_clean['role'] == 'ACTOR'
    ]
    .groupby('id')['name']
    .apply(format_actors)
    .reset_index()
    .rename(
        columns={'name': 'actor'}
    )
)


# --------------------------------------------------
# Merge talent information
# --------------------------------------------------

top_rated = top_rated.merge(
    directors,
    on='id',
    how='left'
)

top_rated = top_rated.merge(
    actors,
    on='id',
    how='left'
)

top_rated['director'] = (
    top_rated['director']
    .fillna('Unknown')
)

top_rated['actor'] = (
    top_rated['actor']
    .fillna('Unknown')
)


# --------------------------------------------------
# Sort for horizontal bar chart
# --------------------------------------------------

top_rated = top_rated.sort_values(
    'imdb_score',
    ascending=True
)


# --------------------------------------------------
# Create chart
# --------------------------------------------------

fig, ax = plt.subplots(
    figsize=(14, 8)
)

bars = ax.barh(
    top_rated['title'],
    top_rated['imdb_score'],
    height=0.6
)


# --------------------------------------------------
# Add labels
# --------------------------------------------------

for bar, score, votes, director, actor in zip(
    bars,
    top_rated['imdb_score'],
    top_rated['imdb_votes'],
    top_rated['director'],
    top_rated['actor']
):

    y = (
        bar.get_y()
        + bar.get_height() / 2
    )

    ax.text(
        score + 0.05,
        y + 0.10,
        f'IMDb: {score:.1f}  |  {votes:,.0f} votes',
        va='center',
        fontsize=9,
        fontweight='bold'
    )

    ax.text(
        score + 0.05,
        y - 0.12,
        f'Director: {director}  |  Actors: {actor}',
        va='center',
        fontsize=8.5
    )


# --------------------------------------------------
# Formatting
# --------------------------------------------------

ax.set_title(
    'Top-Rated Amazon Prime Titles',
    fontsize=18,
    fontweight='bold',
    pad=18
)

ax.text(
    0,
    1.01,
    f'Top 10 titles with at least {MIN_VOTES:,} IMDb votes',
    transform=ax.transAxes,
    fontsize=11
)

ax.set_xlabel(
    'IMDb Score',
    fontsize=12
)

ax.set_ylabel(
    'Title',
    fontsize=12
)

ax.set_xlim(
    max(
        0,
        top_rated['imdb_score'].min() - 0.5
    ),
    10
)

ax.grid(
    axis='x',
    linestyle='--',
    linewidth=0.8,
    alpha=0.25
)

ax.grid(
    axis='y',
    visible=False
)

ax.set_axisbelow(True)

sns.despine(
    left=True,
    bottom=True
)

ax.tick_params(
    axis='both',
    labelsize=9
)

plt.tight_layout()
plt.show()
````

#### Chart - 20 - Bar chart

````python
votes_df = titles_df[
    ['imdb_score', 'imdb_votes']
].copy()

votes_df['imdb_score'] = pd.to_numeric(
    votes_df['imdb_score'],
    errors='coerce'
)

votes_df['imdb_votes'] = pd.to_numeric(
    votes_df['imdb_votes'],
    errors='coerce'
)

votes_df = votes_df.dropna()

# Create meaningful vote-count groups
votes_df['vote_range'] = pd.cut(
    votes_df['imdb_votes'],
    bins=[0, 1000, 10000, 100000, 1000000, float('inf')],
    labels=[
        'Below 1K',
        '1K–10K',
        '10K–100K',
        '100K–1M',
        'Above 1M'
    ]
)

# Calculate average IMDb score for each group
avg_scores = (
    votes_df.groupby(
        'vote_range',
        observed=False
    )['imdb_score']
    .mean()
)

plt.figure(figsize=(10, 6))

bars = plt.bar(
    avg_scores.index,
    avg_scores.values
)

for bar, value in zip(bars, avg_scores.values):
    plt.text(
        bar.get_x() + bar.get_width() / 2,
        bar.get_height() + 0.05,
        f'{value:.2f}',
        ha='center',
        fontsize=10
    )

plt.title(
    'Average IMDb Score by IMDb Vote Range',
    fontsize=18,
    fontweight='bold',
    pad=18
)

plt.xlabel('IMDb Vote Range')
plt.ylabel('Average IMDb Score')

plt.ylim(0, 10)

plt.grid(
    axis='y',
    linestyle='--',
    alpha=0.25
)

sns.despine(left=True, bottom=True)

plt.tight_layout()
plt.show()
````

##### 1. Why did you pick the specific chart?

A bar chart was selected to compare the average IMDb score across different vote-count ranges. Grouping the titles into ranges makes the overall relationship much easier to understand than displaying thousands of individual points.

##### 2. What is/are the insight(s) found from the chart?

The chart shows how average IMDb ratings differ between titles with different levels of audience voting. It helps identify whether titles with greater audience participation tend to have higher or lower average ratings.

##### 3. Will the insights help create a positive business impact?

Yes. These insights can help Amazon understand how audience engagement relates to content ratings and support better content evaluation and promotion strategies. However, IMDb votes represent external audience activity and should not be treated as Amazon's actual viewing data.
